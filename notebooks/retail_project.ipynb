# KPI

```python
# KPI 1: Total Retail Sales
total_retail_sales = df['RETAIL SALES'].sum()

# KPI 2: Total Retail Transfers
total_retail_transfers = df['RETAIL TRANSFERS'].sum()

# KPI 3: Total Warehouse Sales
total_warehouse_sales = df['WAREHOUSE SALES'].sum()

# KPI 4: Number of Suppliers
total_suppliers = df['SUPPLIER'].nunique()

# KPI 5: Number of Products
total_products = df['ITEM CODE'].nunique()

print("Total Retail Sales:", total_retail_sales)
print("Total Retail Transfers:", total_retail_transfers)
print("Total Warehouse Sales:", total_warehouse_sales)
print("Number of Suppliers:", total_suppliers)
print("Number of Products:", total_products)
```

```python
sales_summary = pd.Series({
    'Retail Sales': df['RETAIL SALES'].sum(),
    'Retail Transfers': df['RETAIL TRANSFERS'].sum(),
    'Warehouse Sales': df['WAREHOUSE SALES'].sum()
})

sales_summary
```

```python
import matplotlib.pyplot as plt

ax = sales_summary.plot(
    kind='bar',
    figsize=(8, 5),
    title='Sales & Transfers Comparison'
)

plt.ylabel('Amount')
plt.xlabel('Category')
plt.xticks(rotation=0)

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f')

plt.show()
```

```python
ax = monthly_sales.plot(
    kind='line',
    marker='o',
    figsize=(8, 5),
    title='Monthly Retail Sales Trend'
)

plt.xlabel('Month')
plt.ylabel('Retail Sales')
plt.xticks(monthly_sales.index)
plt.grid()

# Add data labels
for x, y in zip(monthly_sales.index, monthly_sales.values):
    ax.annotate(
        f'{y:,.2f}',
        (x, y),
        textcoords='offset points',
        xytext=(0, 8),
        ha='center'
    )

plt.show()
```

```python
item_sales = (
    df.groupby('ITEM TYPE')['RETAIL SALES']
    .sum()
    .sort_values(ascending=False)
)
```

```python
ax = item_sales.plot(
    kind='barh',
    figsize=(9, 6),
    title='Retail Sales by Item Type'
)

plt.xlabel('Retail Sales')
plt.ylabel('Item Type')

# Add data labels
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)

plt.show()
```

```python
ax = top_products.sort_values().plot(
    kind='barh',
    figsize=(10, 6),
    title='Top 10 Products by Retail Sales'
)

plt.xlabel('Retail Sales')
plt.ylabel('Product')

# Add data labels
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)

plt.show()
```

```python
ax = top_suppliers.sort_values().plot(
    kind='barh',
    figsize=(10, 6),
    title='Top 10 Suppliers by Retail Sales'
)

plt.xlabel('Retail Sales')
plt.ylabel('Supplier')

# Add data labels
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)

plt.show()
```

```python
ax = top_supplier_type.plot(
    kind='bar',
    figsize=(12, 6),
    title='Top 10 Suppliers by Item Type'
)

plt.xlabel('Supplier')
plt.ylabel('Retail Sales')
plt.xticks(rotation=45, ha='right')

# Add data labels
for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=2)

plt.legend(title='Item Type')
plt.tight_layout()
plt.show()
```

```python
product_sales = df.groupby('ITEM DESCRIPTION')[
    ['RETAIL SALES', 'RETAIL TRANSFERS', 'WAREHOUSE SALES']
].sum()

product_sales.head()
```

```python
plt.figure(figsize=(9, 6))

plt.scatter(
    product_sales['RETAIL SALES'],
    product_sales['WAREHOUSE SALES'],
    alpha=0.5
)

plt.title('Retail Sales vs Warehouse Sales by Product')
plt.xlabel('Retail Sales')
plt.ylabel('Warehouse Sales')

plt.grid()
plt.show()
```

```python
item_sales_pct = (
    item_sales / item_sales.sum() * 100
).sort_values(ascending=False)

item_sales_pct
```

```python
ax = item_sales_pct.plot(
    kind='barh',
    figsize=(9, 6),
    title='Retail Sales Contribution by Item Type'
)

plt.xlabel('Contribution (%)')
plt.ylabel('Item Type')

for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', padding=3)

plt.tight_layout()
plt.show()
```

```python
avg_retail_sales_per_product = (
    df.groupby('ITEM DESCRIPTION')['RETAIL SALES']
    .sum()
    .mean()
)

print("Average Retail Sales per Product:", avg_retail_sales_per_product)
```

```python
product_sales_stats = (
    df.groupby('ITEM DESCRIPTION')['RETAIL SALES']
    .sum()
    .agg(['mean', 'median', 'min', 'max'])
)

product_sales_stats
```

```python
zero_sales_products = (
    df.groupby('ITEM DESCRIPTION')['RETAIL SALES']
    .sum()
    .eq(0)
    .sum()
)

total_products = df['ITEM DESCRIPTION'].nunique()

print("Products with zero retail sales:", zero_sales_products)
print("Total unique products:", total_products)
print("Percentage:", zero_sales_products / total_products * 100)
```
